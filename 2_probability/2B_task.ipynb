Tasks: We are going to experiment and explore the expected payout of betting on positions on a roulette wheel.
The highest potential payout position in one go on a roulette wheel is to bet on a single number (assuming that you win). However, the its chances of landing are 1/38. so we have a 37/38 chance of not landing there on a random assumed scenario.

```python
numbers = list(range(1, 39))
numbers
```

Lets say we bet on a random number. Im gonna go with 9 for example.

```python
import random
bet = 9
# To calculate the probability of this bet winning.
count = 0
for n in range(1, 100000):
    res = random.choice(numbers)
    if res == bet:
        count += 1

prob = count/1000
prob
```

So as we can see, over 100000 runs. the chances of a 9 or any single number bet is 1/38 or about 2.63%

Now lets go on the casinos side and see what they are making. and we will look at the expected payout this time and not just the probability.
again i will be loyal to my bet and keep it 9 since changing the number won't do any better to my odds of winning.

The expected value formula is:

$\mathbb{E}[X] = \sum {x \times p(x) } $



```python
# the payout for the casino would be -35 on the player winning and +1 on the player losing. with 37/38 odds of player losing

def expected_payout(n):
    payout = 0
    for x in range(1, n+1):
        expected = (-35*(1/38)) + (1*(37/38))
        payout += expected
        print(payout)


expected_payout(10) # expected payout for the casino after 10 bets. and that's assuming the player only bets 1 dollar each bet.
```

```python
# But what if the player bets 500 dollars on each bet, goes crazy and makes 20 bets. Prolly had a bit too much booz.

def expected_payout(n):
    payout = 0
    for x in range(1, n+1):
        expected = (-(500*35) * (1/38)) + (500*(37/38))
        payout += expected
        print(payout)


expected_payout(20)
```

```python
# i wanna try to visualise the rate at which it grows.
import matplotlib.pyplot as plt
def expected_payout(n):
    payout = []
    tally = 0
    for x in range(1, n+1):
        expected = (-35 * (1/38)) + (1*(37/38))
        tally += expected
        payout.append(tally)

    return payout

payout = expected_payout(10000)
plt.plot(payout)
plt.legend()
```
